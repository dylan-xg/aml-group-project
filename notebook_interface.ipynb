# Notebook Interface


A notebook for testing the video feed since it supports using jupyter widgets for display.


```python
import os

import cv2 as cv

from src.panopticon.settings import SETTINGS
from src.panopticon import video_processor as VP
```

```python
FORCE_VID = True

if not SETTINGS.TESTING_VID:
	raise ValueError('`TESTING_VID` setting not set.')

if not SETTINGS.TESTING_VID.exists():
	raise ValueError(f'File not found: {SETTINGS.TESTING_VID}')

# `VIDEO_SOURCE` will display webcam when run locally, or video file when run remotely.
if 'SSH_CLIENT' in os.environ:
	VIDEO_SOURCE = SETTINGS.TESTING_VID
else:
	VIDEO_SOURCE = SETTINGS.TESTING_VID if FORCE_VID else 0
```

```python
def callback_function(frame: cv.typing.MatLike) -> cv.typing.MatLike:
	return cv.cvtColor(src=frame, code=cv.COLOR_RGB2GRAY)
```

```python
# Forcing video file for testing currently.
VP.process_video(
	capture_location=VIDEO_SOURCE,
	callback=callback_function,
	display_config=VP.DisplayConfig.Jupyter(frametime=1/60)
)
```
