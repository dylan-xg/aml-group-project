# Notebook Interface


A notebook for testing the video feed since it supports using jupyter widgets for display.


```python
import os
from pathlib import Path

import cv2 as cv

from src.panopticon.video_processor import DisplayConfig, VideoFeed
```

```python
# You will need to add your own video here
PRERECORDING_PATH = Path('data/testing/example.mp4')

# `VIDEO_SOURCE` will display webcam when run locally, or video file when run remotely.
if 'SSH_CLIENT' in os.environ:
	print('Remote session detected. Using video file.')
	VIDEO_SOURCE = PRERECORDING_PATH
else:
	print('Local session detected. Using live webcam.')
	VIDEO_SOURCE = 0
```

```python
def callback_function(frame: cv.typing.MatLike) -> cv.typing.MatLike:
	return cv.cvtColor(src=frame, code=cv.COLOR_RGB2GRAY)
```

```python
# Forcing video file for testing currently.
VideoFeed.process_video(
	capture_location=PRERECORDING_PATH,
	callback=callback_function,
	display_config=DisplayConfig.Jupyter(frametime=1/60)
)
```
